**Answer**

- REST stands for Representational State Transfer and API stands for Application Program Interface. REST is a software architectural style that defines the set of rules to be used for creating web services. Web services that follow the REST architectural style are known as RESTful web services. It allows requesting systems to access and manipulate web resources by using a uniform and predefined set of rules. Interaction in REST-based systems happens through the Internet's Hypertext Transfer Protocol (HTTP).

- A Restful system consists of a:

   - A client who requests for the resources.
   - server who has the resources.

- It is important to create REST API according to industry standards which results in ease of development and increase client adoption.

- **Architectural Constraints of RESTful API**

- There are six architectural constraints that makes any web service are listed below:

   1. **Uniform Interface**
     - It is a key constraint that differentiates between a REST API and a Non-REST API. It suggests that there should be a uniform way of interacting with a given server irrespective of device or type of application (website, mobile app).

     - There are four guidelines principles of a Uniform Interface are:

        - **Resource-Based**: Individual resources are identified in requests. For example: API/users.
        - **Manipulation of Resources Through Representations:** The client has a representation of the resource and it contains enough information to modify or delete the resource on the server, provided it has permission to do so. Example: Usually user gets a user ID when the user requests a list of users and then uses that ID to delete or modify that particular user.
        - **Self-descriptive Messages:** Each message includes enough information to describe how to process the message so that the server can easily analyze the request.
        - **Hypermedia as the Engine of Application State (HATEOAS):** It need to include links for each response so that client can discover other resources easily.
  2. **Stateless**
    - It means that the necessary state to handle the request is contained within the request itself and server would not store anything related to the session. In REST, the client must include all information for the server to fulfill the request whether as a part of query params, headers or URI. Statelessness enables greater availability since the server does not have to maintain, update or communicate that session state. There is a drawback when the client need to send too much data to the server so it reduces the scope of network optimization and requires more bandwidth.

  3. **Cacheable**
   - Every response should include whether the response is cacheable or not and for how much duration responses can be cached at the client side. Client will return the data from its cache for any subsequent request and there would be no need to send the request again to the server. A well-managed caching partially or completely eliminates some client–server interactions, further improving availability and performance. But sometime there are chances that user may receive stale data.

   4. **Client-Server**
    - REST application should have a client-server architecture. A Client is someone who is requesting resources and are not concerned with data storage, which remains internal to each server, and server is someone who holds the resources and are not concerned with the user interface or user state. They can evolve independently. Client doesn't need to know anything about business logic and server doesn't need to know anything about frontend UI.

   5. **Layered system**
    - An application architecture needs to be composed of multiple layers. Each layer doesn't know any thing about any layer other than that of immediate layer and there can be lot of intermediate servers between client and the end server. Intermediary servers may improve system availability by enabling load-balancing and by providing shared caches.

   6. **Code on demand**
   - It is an optional feature. According to this, servers can also provide executable code to the client. The examples of code on demand may include the compiled components such as Java Servlets and Server-Side Scripts such as JavaScript.

- **GET Method**
   - The HTTP GET method retrieves a resource. On success, it returns data (usually JSON or XML) with 200 OK, and on error, it commonly returns 404 Not Found or 400 Bad Request.

    - GET /users/123
      This request fetches data for the user with ID 123.

- **POST Method**
   - The POST method creates new resources. On success, it returns 201 Created with a Location header pointing to the new resource.

   - POST /users

     {
        "name": "Anne",
        "email": "gfg@example.com"
      }
   - This request creates a new user with the given data.

- **PUT Method**
   - PUT is used to update or create a resource. It sends the complete resource in the request body and replaces the existing one at the specified URL.

  - PUT /users/123

    {
       "name": "Anne",
       "email": "gfg@example.com"
    }
   - This request updates the user with ID 123 or creates a new user if one doesn't exist.

- **DELETE Method**
  - It is used to delete a resource identified by a URI. On successful deletion, return HTTP status 200 (OK) along with a response body.

  - DELETE /users/123
   This request deletes the user with ID 123.

**Question 3:** Add validation to:
- course_id must be integer > 0
- level must be one of: beginner, intermediate, advanced

What happens if validation fails?

```python
# ANSWER
from fastapi import FastAPI, Query, Path, HTTPException
from pydantic import BaseModel
from typing import List, Optional
from enum import Enum

app = FastAPI(
    title="Course API",
    description="A simple FastAPI application demonstrating endpoints, path parameters, and query parameters",
    version="1.0.0"
)

# Enum for strict level validation
class CourseLevel(str, Enum):
    BEGINNER = "beginner"
    INTERMEDIATE = "intermediate"
    ADVANCED = "advanced"

class Course(BaseModel):
    id: int
    title: str
    level: CourseLevel
    description: Optional[str] = None

courses_db = [
    Course(id=1, title="Engg Mathamatics", level=CourseLevel.BEGINNER, description="Learn engg maths"),
    Course(id=2, title="C++", level=CourseLevel.INTERMEDIATE, description="Learn c++"),
    Course(id=3, title="Advanced python", level=CourseLevel.ADVANCED, description="Learn python"),
    Course(id=4, title="Data Science", level=CourseLevel.INTERMEDIATE, description="Analytics and ML basics"),
]


@app.get("/")
def read_root():
    return {
        "message": "Welcome to the Course API!",
        "available_endpoints": {
            "GET /courses": "Get all courses (with optional level filter)",
            "GET /courses/{course_id}": "Get a specific course by ID"
        }
    }


@app.get("/courses", response_model=List[Course])
def get_courses(
    level: Optional[CourseLevel] = Query(
        None,
        description="Filter courses by level (beginner, intermediate, advanced)"
    )
):
    """Get all courses with optional level filter"""
    if level:
        filtered_courses = [
            course for course in courses_db
            if course.level == level
        ]
        return filtered_courses

    return courses_db


@app.get("/courses/{course_id}", response_model=Course)
def get_course(
    course_id: int = Path(..., gt=0, description="Course ID must be a positive integer (> 0)")
):
    """Get a specific course by ID"""
    for course in courses_db:
        if course.id == course_id:
            return course

    raise HTTPException(
        status_code=404,
        detail=f"Course with ID {course_id} not found"
    )
```

 **schema validation is critical in APIs.**
- Schema validation is critical in APIs because it acts as the first line of defense and a foundational contract for reliable, secure, and maintainable communication between clients and servers.

following reasons:

1. **Data Integrity & Correctness**
- Ensures that the data received by your API matches the expected structure, types, formats, and constraints (e.g., price >= 0, duration > 0, required fields, email format, etc.).
- Prevents bad or incomplete data from entering your business logic, database, or downstream services.

2. **Security**
- Blocks malicious inputs such as SQL injection, NoSQL injection, oversized payloads, unexpected fields, or type coercion attacks.
- Helps mitigate mass assignment vulnerabilities (where attackers send extra fields to overwrite sensitive data).
- Enforces strict rules on sensitive fields (e.g., password length, role permissions).

3. **Early Error Detection**
- Catches invalid requests immediately at the API boundary instead of deep inside the application.
- Returns clear, structured error messages (e.g., "price must be greater than or equal to 0") instead of cryptic 500 errors.
- Improves debugging and reduces the blast radius of bugs.

4. **API Contract & Documentation**
- The schema (OpenAPI/Swagger, Pydantic models, Zod, etc.) serves as a living contract between frontend, backend, and third-party consumers.
- Tools can auto-generate documentation, client SDKs, and type definitions from the schema.
- Makes integration much smoother and reduces "it works on my machine" issues.

5. **Consistency & Maintainability**
- Enforces uniform data formats across your entire system.
- Makes it easier to evolve the API (deprecate fields, add new ones) with confidence.
- Simplifies testing, you can reliably test business logic assuming valid data.

6. **Performance**
- Invalid requests are rejected very early (often before hitting business logic or database), saving server resources.
- Reduces unnecessary processing of malformed data.

**Question 5:** Explain:
- Authentication vs Authorization
- JWT-based authentication
- Why stateless authentication is scalable

**Answer**

- **Authentication vs Authorization**
   - Authentication and Authorization are key security steps that ensure only the right users access a system and can perform permitted actions. Authentication verifies who you are, and Authorization determines what you can do.

      - Authentication: Confirms the user’s identity.
      - Authorization: Determines what the verified user is allowed to access or perform.

- **Authentication**

   - Authentication is the process of verifying the identity of a user or system. It ensures that the user is legitimate by validating credentials like passwords, OTPs, or biometrics.

   - **Working**

   - The authentication process verifies the identity of a user before granting access to the system.

      - User enters credentials (password, OTP, biometrics)
      - System verifies the credentials
      - If valid, the user is successfully authenticated
- **Authorization**
    - Authorization determines the access rights and permissions of an authenticated user. It decides what resources the user can access and what actions they are allowed to perform.

    - **Working**

    - The system uses authentication and authorization processes to control access and ensure security.

       - System checks the user’s roles or permissions
       - Grants or denies access to resources
       - Ensures the user can perform only allowed actions

- **JWT-based authentication**
- JSON Web Token (JWT) is a compact and secure method for transmitting information between parties as a JSON object. It is commonly used for authentication and authorization in web applications, allowing users to access protected resources without repeatedly providing credentials. Key features include:

   - **Stateless Authentication:** Stores user information in a token, reducing the need for server-side session storage.
   - Secure Data Exchange: Uses digital signatures to verify that the token has not been altered.
   - **Compact Format:** Consists of three parts—Header, Payload, and Signature—encoded into a single string.
   - **Widely Supported:** Works across different programming languages and platforms.

- **JWT Structure**
- JWT mainly consists of 3 components, separated by dots(.):

Header. Payload. Signature

1. **Header**
- The header contains metadata about the token, including the signing algorithm and token type.

{
    "alg": "HS256",
    "typ": "JWT"
}

where,

alg: Algorithm used for signing (e.g., HS256, RS256).

typ: Token type, always "JWT".

2. **Payload**

- The payload contains the information about the user also known as a claim and some additional information including the timestamp at which it was issued and the expiry time of the token.

{
    "userId": 123,
    "role": "admin",
    "exp": 1672531199
}


3. **Signature**
- The signature ensures token integrity and is generated using the header, payload, and a secret key.

HMACSHA256(

    base64UrlEncode(header) + "." + base64UrlEncode(payload),
    secret
)


**stateless authentication is scalable:**

- following reasons:

1. **No Centralized State**
- You don’t need a shared Redis, database, or session store that becomes a bottleneck.
Every server instance is completely independent.

2. **Horizontal Scaling Made Easy**
- You can add more servers (or containers) instantly.
- Any incoming request can be handled by any server in your cluster without coordination.

3. **Reduced Latency & Higher Throughput**
- Validating a JWT is just a cryptographic signature check (very fast).
- No network call to Redis or database for session lookup.

4. **Better for Distributed Systems & Microservices**
- Each service can independently verify the token.
- No need to call an "auth service" for every request.

5. **Easier Caching & CDN Usage**
Static content + authenticated routes become easier to manage.
Edge computing (Cloudflare, Fastly, etc.) can sometimes handle token validation.

6. **Fault Tolerance**
If one server goes down, others don’t lose user sessions.

**API Testing is Critical**
1. **Ensures Reliability & Correctness**

- Verifies that your API endpoints return the expected responses for given inputs.

4. **Supports Scalability & Performance**

- Identifies bottlenecks under load (e.g., slow database queries, inefficient authentication).

**Difference Between Unit and Integration Testing**

**Unit Testing:** Unit Testing is a type of software testing where individual software components are tested. Unit Testing of the software product is carried out during the development of an application. An individual component may be either an individual function or a procedure. Unit Testing is typically performed by the developer. It is a testing method using which every independent module are tested to determine if there is any issue by the developer himself.

**Integration Testing:** Integration testing is the process of testing the interface between two software units or modules. Its focus is on determining the correctness of the interface. Integration testing aims to expose faults in the interaction between integrated units. Once all the modules have been unit tested, integration testing is performed.


```python
#main.py (Your FastAPI app)
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel

app = FastAPI()

class Item(BaseModel):
    name: str
    price: float

items = {}

@app.post("/items/", status_code=201)
async def create_item(item: Item):
    if item.name in items:
        raise HTTPException(status_code=400, detail="Item already exists")
    items[item.name] = item
    return {"message": "Item created", "item": item}

@app.get("/items/{name}")
async def get_item(name: str):
    if name not in items:
        raise HTTPException(status_code=404, detail="Item not found")
    return items[name]
```

**Answer**


These are the three main ways to send data to a web API endpoint. They differ in where the data is placed, how visible it is, HTTP method suitability, and intended purpose.

1. **Path Parameter**

**Definition**: A variable part of the URL path that identifies a specific resource.

**Example**:

GET /users/42
GET /posts/abc-123/comments/5


**FastAPI Example**:

from fastapi import FastAPI

app = FastAPI()

@app.get("/users/{user_id}")

def get_user(user_id: int):     
      # Path parameter
    return {"user_id": user_id}

**When to use**:
- To **identify a single resource** (most common use).
- For hierarchical resources (`/users/{user_id}/orders/{order_id}`).
- When the value is **essential** to locate the resource.
- For **RESTful** resource-oriented design.

**Best practices**:
- Keep them simple and clean.
- Use meaningful names.
- Prefer UUIDs over plain integers in production for security.



2.**Query Parameter**

**Definition**: Key-value pairs appended to the URL after a question mark `?`.

**Example**:

GET /users?role=admin&active=true&limit=20&offset=0
GET /products?category=electronics&min_price=100&sort=price


**FastAPI Example**:

@app.get("/users")
def get_users(
    role: str | None = None,      # Query parameter
    active: bool = True,
    limit: int = 100,
    offset: int = 0
):
    


**When to use**:
- **Filtering**, searching, sorting, and pagination.
- Optional parameters.
- When you want the request to be **cacheable** (GET requests).
- For **lightweight** data that modifies the behavior of the endpoint.

**Best practices**:
- Use clear, descriptive names.
- Provide sensible defaults.
- Consider using query params for complex filtering instead of request body on GET requests.



3. **Request Body**

**Definition**: Data sent in the body of the HTTP request, typically as JSON.

```python
#ANSWER

# 1.BaseModel

#This is the foundation. All your data models inherit from it.


from pydantic import BaseModel

class UserCreate(BaseModel):
    name: str
    email: str
    age: int | None = None
    is_active: bool = True


#Key features:
'''
- Automatic validation of input data
- Type conversion (e.g., string → int)
- `.dict()`, `.json()`, `.model_dump()` methods
- Used heavily in FastAPI for request/response bodies'''


#2.Field
#Used to add extra information or constraints to individual fields.

from pydantic import BaseModel, Field

class UserCreate(BaseModel):
    name: str = Field(..., min_length=2, max_length=50)
    email: str = Field(..., pattern=r'^.+@.+\..+$', description="User email address")
    age: int = Field(ge=0, le=120, default=None, title="Age")
    password: str = Field(..., min_length=8, exclude=True)  # not returned in .dict()


#Common parameters:
'''
- `default`, `default_factory`
- `min_length`, `max_length`, `gt`, `ge`, `lt`, `le`, `pattern`
- `description`, `title`, `alias`, `exclude`'''


#3.validator(Field Validator)
#Runs custom validation logic on one specific field
from pydantic import BaseModel, validator

class UserCreate(BaseModel):
    name: str
    email: str

    @validator('name')
    def name_must_not_be_empty(cls, v: str) -> str:
        if not v or not v.strip():
            raise ValueError('Name cannot be empty')
        return v.title()

    @validator('email')
    def validate_email(cls, v: str) -> str:
        if '@' not in v:
            raise ValueError('Invalid email')
        return v.lower()

'''
- Runs **after** type validation.
- Receives the value of that field.
- Can transform the value or raise `ValueError`.'''


# 4.root_validator(Model Validator)

#Runs validation on the **entire model** (multiple fields at once).


from pydantic import BaseModel, root_validator

class UserCreate(BaseModel):
    password: str
    password_confirm: str
    age: int
    is_student: bool = False

    @root_validator(pre=True)   # or skip `pre=True`
    def check_passwords_match(cls, values: dict):
        if values.get('password') != values.get('password_confirm'):
            raise ValueError('Passwords do not match')
        return values

    @root_validator()
    def check_age(cls, values: dict):
        if values.get('is_student') and values.get('age', 0) > 30:
            raise ValueError('Students over 30? Suspicious!')
        return values

'''
- Useful for **cross-field validation**.
- `pre=True` runs before field validation.
- Receives a dictionary of all values.
'''


```

Why should secrets never be hardcoded?


```python
from pydantic_settings import BaseSettings


class Settings(BaseSettings):
    """Application configuration settings."""

    # Database
    DATABASE_URL: str

    # Security
    SECRET_KEY: str

    # Development
    DEBUG: bool = False

    class Config:
        # Load environment variables from .env file
        env_file = ".env"
        env_file_encoding = "utf-8"



# Create a global settings instance
settings = Settings()
```

**Why should secrets never be hardcoded ?**
1. **Security Risk (Biggest Reason)**

- If your secret is in the code, anyone who sees your code can use it.
- This includes:
   - Teammates
   - Open-source contributors
   - Anyone who finds your notebook on GitHub
   - Hackers scraping public repositories


2. **Version Control Exposure**

- Hardcoded secrets get committed to Git history.
- Even if you delete it later, it can still be recovered from Git history.
- Many companies have had major breaches because of secrets leaked on GitHub.

3. **Real-World Consequences**

- Database credentials leaked → attackers can steal or destroy data.
- SECRET_KEY leaked → session hijacking, JWT forgery, etc.
- API keys leaked → huge cloud bills or data breaches.
